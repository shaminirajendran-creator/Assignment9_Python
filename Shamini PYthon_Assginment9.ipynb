```python
import random

secret_number = random.randint(1, 10)
attempts = 3
print("Welcome to the Number Guessing Game!")
print("Guess the secret number between 1 and 10.")
print("You have", attempts, "attempts.")
```

```python
while attempts > 0:
    guess = int(input("Enter your guess: "))

    # Check if guess is out of range
    if guess < 1 or guess > 10:
        print("Invalid guess! Please enter a number between 1 and 10.")
        continue  # Skip rest of loop and ask again

    # Check guess against secret number
    if guess == secret_number:
        print("Congratulations! You guessed the correct number:", secret_number)
        break  # Exit loop if correct
    elif guess > secret_number:
        print("Too high! Try again.")
    else:
        print("Too low! Try again.")

    # Reduce attempts
    attempts -= 1
    print("Attempts left:", attempts)

else:
    # Executed if loop ends without break
    print("Better luck next time! The secret number was:", secret_number)
```

```python
number = int(input("Enter a number to generate its multiplication table: "))
```

```python
print(f"\nMultiplication Table for {number}:")
for i in range(1, 10 + 1):   # range from 1 to 10
    result = number * i
    print(f"{number} x {i} = {result}")
```

```python
# Define the BMI calculation function
def calculate_bmi(weight, height):
    return weight / (height ** 2)

# Prompt the user for input
weight = float(input("Enter your weight in kilograms: "))
height = float(input("Enter your height in meters: "))

# Calculate BMI
bmi = calculate_bmi(weight, height)

# Display the result
print(f"Your BMI is: {bmi:.2f}")
```
