```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
#Inspection and Properties
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

```python
#Array Operations

#Convert Celsius → Fahrenheit:

temperatures_f = (temperatures_w1 * 9/5) + 32
print("Fahrenheit:", temperatures_f)
```

```python
#Find statistics:

print("Max:", np.max(temperatures_w1))
print("Min:", np.min(temperatures_w1))
print("Mean:", np.mean(temperatures_w1))
```

```python
#Array Slicing and Indexing
# First three days
print("First 3 days:", temperatures_w1[:3])

# Weekend (last two days)
print("Weekend:", temperatures_w1[-2:])

# Middle three days (index 2,3,4)
print("Middle 3 days:", temperatures_w1[2:5])
```

```python
#Create a 2D Array

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],  # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]   # Week 2
])
print(temperatures)
```

```python
#Pandas Series
#1. Creating Pandas Series
import pandas as pd

marks = pd.Series([95, 92, 89, 85, 80],
                  index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)


```

```python
#2. Indexing and Slicing
# a. Integer index position (1st rank student → index 0)
print("1st rank mark:", marks[0])

# b. loc accessor (top 3 ranks by labels)
print("Top 3 ranks (loc):")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# c. iloc accessor (3rd rank student → index 2)
print("3rd rank mark (iloc):", marks.iloc[2])

# d. Boolean mask (marks > 90)
print("Marks > 90:")
print(marks[marks > 90])
```

```python
#3. Manipulating Series
# a. Modify 1st rank mark to 100
marks['Rank1'] = 100
print("Updated marks:\n", marks)

# b. Remove last rank entry
marks = marks.drop('Rank5')
print("After removing last rank:\n", marks)

# c. Compute CGPA (marks ÷ 10)
cgpa = marks / 10
print("CGPA:\n", cgpa)
```

```python
# Pandas DataFrame

import pandas as pd

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East',
               'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
print(transactions)
```

```python
#2. Data Exploration
# Display DataFrame
print(transactions)

# Basic info
print(transactions.head())       # First 5 rows
print(transactions.tail())       # Last 5 rows
print("Shape:", transactions.shape)
print("Columns:", transactions.columns)
print("Data types:\n", transactions.dtypes)

# Display specific columns
print(transactions[['ProductCategory', 'Amount']])

# Retrieve last 3 columns
print(transactions.iloc[:, -3:])   # Using iloc

# Filter rows (Region = North & Amount > 200)
print(transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])

# Value counts for ProductCategory
print(transactions['ProductCategory'].value_counts())

# Unique values in Region
print(transactions['Region'].unique())

# Group by Region → mean Amount
print(transactions.groupby('Region')['Amount'].mean())
```

```python
#3. Manipulating the DataFrame
# Modify Amount for TransactionID 102 → 165
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions)

# Add Discount column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)

# Remove row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]
print(transactions)

# Delete Discount column
transactions = transactions.drop(columns=['Discount'])
print(transactions)
```
